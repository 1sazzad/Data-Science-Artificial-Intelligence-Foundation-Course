# **The Main Types of Seaborn Plots**  
Think of Seaborn plots in 4 main categories.
| Category           | Purpose                      |
| ------------------ | ---------------------------- |
| Distribution plots | Understand data distribution |
| Relationship plots | Relation between variables   |
| Categorical plots  | Compare categories           |
| Matrix plots       | Correlation / heatmaps       |


## **Distribution plots**  
`histplot()` `kdeplot()` `displot()` `ecdfplot()` `rugplot()`


### **Histogram - `histplot()`**  
A histogram divides data into bins and counts frequency.  
$$
\text{frequency} = \text{count}(x_i \in \text{bin}_j)
$$


````python
# Histplot()

import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(tips["total_bill"])

plt.show()
````

````python
# Control Number of Bins
sns.histplot(tips["total_bill"], bins=30)
````

````python
# Add KDE Curve
sns.histplot(tips["total_bill"], kde=True)
````

### **KDE Plot (Kernel Density Estimation)**  
Histogram is discreat.  
KDE estimates **continuous probability density**  
Formula:  
$$
f(x) = \frac{1}{nh} \sum_{i=1}^{n} K\left(\frac{x - x_i}{h}\right)
$$
Where:  
* K = kernel function
* h = bandwidth
* n = observations

````python
sns.kdeplot(tips["total_bill"])
plt.show()
````

````python
# Multiple Groups
sns.kdeplot(data=tips, x="total_bill", hue="sex")
````

### **Displot (Figure Level)**  
`displot()` is a high-level distribution function. It can automatically create multiple distribution plots.

````python
sns.displot(tips["total_bill"], kde=True)
````

````python
# By Category
sns.displot(data=tips, x="total_bill", col="sex", kde=True)
````

### **ECDF Plot**  
Probability that a variable is less than or equal to x.  
Like: $P(X \le x)$


````python
sns.ecdfplot(tips["total_bill"])
````

````python
sns.kdeplot(tips["total_bill"])
sns.rugplot(tips["total_bill"])
````

## **Relationship Plots**  
Relationship plots help answer questions like:

* Are two variables **positively correlated?**

### **Scatter Plot**  
A scatter plot shows the relationship between two numerical variables.

Mathematically, we visualize pairs:  
$(x_1, y_1), (x_2, y_2), \dots, (x_n, y_n)$  
Each point represents one observation.  
Scatter plot helps us see:  
* correlation  
* clusters  
* outliers  
* patterns

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.show()
````

### **Adding Categories with Color**   
One of Seaborn's strengths is **visual grouping.**  

````python
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips)
````

### **Adding More Dimensions**  
We can visualize **3 variables at once.**  

````python
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",
    data=tips
)
````

### **Regression Plot (regplot)**  
This plot shows **relationship + regression line.**  
Regression equation:   $y = \beta_0 + \beta_1 x$  
Where:  
* $\beta_0$ = intercept
* $\beta_1$ = slope

````python
sns.regplot(x="total_bill", y="tip", data=tips)
````

### **lmplot (Advanced Regression Plot)**

## **Categorical Plots**  
A **categorical variable** is a variable that represents **groups or categories** instead of continuous numbers.  
| Category | Example Values |
| -------- | -------------- |
| Gender   | Male, Female   |
| Day      | Mon, Tue, Wed  |
| Class    | A, B, C        |
| Product  | Phone, Laptop  |

Seaborn provides specialized plots to analyze **numerical values across categories.**

### **Count Plot**  
A **count plot** shows the **frequency of each category.**  
Mathematically:  
$$
C = \{c_1, c_2, \dots, c_k\}
$$


````python
sns.countplot(x="day", data=tips)

plt.show()
````

````python
# With grouping
sns.countplot(x="day", hue="sex", data=tips)
````

### **Bar Plot**  
A bar plot shows the average (mean) of a numerical variable for each category.  
Formula:  
$$
\text{mean} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$


````python
sns.barplot(x="day", y="total_bill", data=tips)
````

````python
# With grouping
sns.barplot(x="day", y="total_bill", hue="sex", data=tips)
````

### **Box Plot ⭐**  
Very important in data science.  
A box plot visualizes distribution using quartiles.  
Five-number summary:  
$$
 \text{min},\; Q_1,\; \text{median},\; Q_3,\; \text{max}
$$
Where  
* **Q1** = 25th percentile  
* **Median** = 50th percentile  
* **Q3** = 75th percentile

Outliers are also shown.

````python
sns.boxplot(x="day", y="total_bill", data=tips)
````

````python
# With grouping
sns.boxplot(x="day", y="total_bill", hue="sex", data=tips)
````

### **Violin Plot**  
A **violin plot** is an advanced version of boxplot.  
It combines:  
* boxplot  
* KDE distribution  

Meaning:  
It shows data distribution shape.

````python
sns.violinplot(x="day", y="total_bill", data=tips)
````

### **Strip Plot**

### **Swarm Plot**  
Improved version of strip plot (avoids overlapping points).

````python
sns.swarmplot(x="day", y="total_bill", data=tips)
````

## **Matrix Plots**  
Matrix plots visualize **relationships between many variables at once.**  
They are mainly used to analyze:
* Correlation between features  
* Similarity between observations
* Feature grouping

The two main plots are:  
1. Heatmap  
2. Clustermap

### **Correlation Matrix**  
Correlation measures the **strength of relationship between two variables.**  
The most common measure is **Pearson Correlation.**  
$$
r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}
{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}
$$  
Where:  
r = 1 → perfect positive correlation  
r = − 1 → perfect negative correlation  
r = 0 → no correlation  
Example correlation table:  
|   | A    | B   | C    |
| - | ---- | --- | ---- |
| A | 1    | 0.8 | -0.2 |
| B | 0.8  | 1   | 0.3  |
| C | -0.2 | 0.3 | 1    |


### **Heatmap**  
A **heatmap** visualizes a matrix using color intensity.  
* Dark color → strong relationship  
* Light color → weak relationship

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips = sns.load_dataset("tips")

# Select only numeric columns for correlation calculation
numeric_tips = tips.select_dtypes(include=np.number)
corr = numeric_tips.corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.show()
````

### **Annotated Heatmap**  
Data scientists usually show actual correlation values.

````python
sns.heatmap(corr, annot=True)
````

````python
sns.heatmap(corr, annot=True, cmap="coolwarm")
````

### **Clustermap**  
A **clustermap** groups similar variables together.  
It performs **hierarchical clustering** automatically.  
Meaning:  
Variables with similar patterns will appear **closer together**.
