# **Temperature Analysis**  
**Goal:** Simulate 30 days of temperature and analyze.  
**Requirements:**
* Line plot
* Scatter
* Histogram
* Subplot
* Mean Line  




````python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------
# Style
# --------------------------------
plt.style.use("ggplot")

# --------------------------------
# Generate Data
# --------------------------------
days = np.arange(1, 31)
temperature = np.random.randint(20, 40, 30)

mean_temp = np.mean(temperature)

# --------------------------------
# Figure Setup
# --------------------------------
fig, axs = plt.subplots(2, 2, figsize=(12, 8))
axs = axs.flatten()

# Common ticks
x_ticks = np.arange(0, 31, 5)
y_ticks = np.arange(20, 41, 5)

# --------------------------------
# Plot Configurations
# --------------------------------
titles = [
    "Temperature Trend",
    "Temperature Scatter",
    "Temperature Distribution",
    "Temperature with Mean"
]

# --------------------------------
# 1️⃣ Line Plot
# --------------------------------
axs[0].plot(days, temperature, marker='o', linewidth=2)

axs[0].set_title(titles[0])
axs[0].set_xlabel("Day")
axs[0].set_ylabel("Temperature (°C)")

axs[0].set_xticks(x_ticks)
axs[0].set_yticks(y_ticks)

axs[0].grid(True, linestyle='--', alpha=0.7)

# --------------------------------
# 2️⃣ Scatter Plot
# --------------------------------
axs[1].scatter(days, temperature, s=80, color='red')

axs[1].set_title(titles[1])
axs[1].set_xlabel("Day")
axs[1].set_ylabel("Temperature (°C)")

axs[1].set_xticks(x_ticks)
axs[1].set_yticks(y_ticks)

axs[1].grid(True, linestyle='--', alpha=0.7)

# --------------------------------
# 3️⃣ Histogram
# --------------------------------
axs[2].hist(temperature, bins=10, edgecolor='black')

axs[2].set_title(titles[2])
axs[2].set_xlabel("Temperature (°C)")
axs[2].set_ylabel("Frequency")

axs[2].set_xticks(np.arange(20, 41, 2))

axs[2].grid(True, linestyle='--', alpha=0.7)

# --------------------------------
# 4️⃣ Line Plot + Mean Line
# --------------------------------
axs[3].plot(days, temperature, marker='o', linewidth=2)

axs[3].axhline(
    mean_temp,
    color='black',
    linestyle='--',
    linewidth=2,
    label=f"Mean = {mean_temp:.2f}"
)

axs[3].set_title(titles[3])
axs[3].set_xlabel("Day")
axs[3].set_ylabel("Temperature (°C)")

axs[3].set_xticks(x_ticks)
axs[3].set_yticks(y_ticks)

axs[3].legend()

axs[3].grid(True, linestyle='--', alpha=0.7)

# --------------------------------
# Layout
# --------------------------------
fig.suptitle("Monthly Temperature Analysis", fontsize=16)

plt.tight_layout()

plt.show()
````

# **Mathematical Function Dashboard**  
**Goal**  
Build a function visualization dashboard.  
**Requirements**  
Plot:
* Linear Function
* Quadratic Function
* Exponential
* Logarithmic

All in one figure.

**Data**  
```
x = np.linspace(1, 10, 400)
y1 = 2*x + 1
y2 = x**2
y3 = np.exp(x/5)
y4 = np.log(x)
```
**What You Must Do**  
1. 2×2 subplot layout
2. Use different colors
3. Add grid
4. Add minor ticks
5. Share x-axis where appropriate
6. Set axis limits properly

````python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Create Dataset
# -----------------------------
x = np.linspace(1, 10, 40)

functions = [
    2*x + 1,
    x**2,
    np.exp(x/5),
    np.log(x)
]

titles = [
    "Linear Function",
    "Quadratic Function",
    "Exponential Function",
    "Logarithmic Function"
]

colors = ['b', 'g', 'r', 'y']

xticks = [
    np.arange(1,11,1),
    np.arange(0,11,1),
    np.arange(0,11,1),
    np.arange(1,11,1)
]

yticks = [
    np.arange(0,22,2),
    np.arange(0,101,10),
    np.arange(0,11,1),
    np.arange(0,2.5,0.2)
]

# -----------------------------
# Create Subplots
# -----------------------------
fig, axs = plt.subplots(2, 2, figsize=(10,8))

axs = axs.flatten()   # Convert 2x2 array → 1D for easy looping

# -----------------------------
# Plot Loop
# -----------------------------
for i in range(4):

    axs[i].plot(x, functions[i], color=colors[i])

    axs[i].set_title(titles[i], fontsize=12)

    axs[i].set_xticks(xticks[i])
    axs[i].set_yticks(yticks[i])

    axs[i].minorticks_on()

    axs[i].grid(True, which='both', linestyle='--', alpha=0.6)

# -----------------------------
# Adjust Layout
# -----------------------------
plt.tight_layout()

plt.show()
````

# **Mini Data Dashboard**  
**Goal**  
Create a professional data dashboard.  
Simulate a small company dataset:  
* Monthly sales
* Monthly expenses
* Profit
* Category distribution  

**Data Example**  

months = np.array(["Jan","Feb","Mar","Apr","May","Jun"])  
sales = np.array([200, 220, 250, 240, 260, 300])  
expenses = np.array([150, 160, 180, 170, 200, 210])  
profit = sales - expenses  
categories = ["Product A", "Product B", "Product C"]  
distribution = np.array([50, 30, 20])  

**What You Must Do**  
Create ONE big figure with:  
1. Line plot → Sales & Expenses
2. Bar plot → Profit
3. Pie chart → Product distribution
4. Histogram → Profit variation  

**Add:**  
* Legend
* Grid
* Proper layout
* Titles
* Clean style  

Make it look like a business report.

````python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Dataset
# -----------------------------
months = np.array(["Jan","Feb","Mar","Apr","May","Jun"])

sales = np.array([200, 220, 250, 240, 260, 300])
expenses = np.array([150, 160, 180, 170, 200, 210])

profit = sales - expenses

categories = ["Product A", "Product B", "Product C"]
distribution = np.array([50, 30, 20])

# -----------------------------
# Create Subplots
# -----------------------------
fig, axs = plt.subplots(2,2, figsize=(10,8))

axs = axs.flatten()

# -----------------------------
# 1️⃣ Line Plot (Sales vs Expenses)
# -----------------------------
axs[0].plot(months, sales, marker='o', label='Sales')
axs[0].plot(months, expenses, marker='o', label='Expenses')

axs[0].set_title("Sales vs Expenses")
axs[0].set_xlabel("Months")
axs[0].set_ylabel("Amount")

axs[0].grid(True, linestyle='--', alpha=0.6)
axs[0].legend()

# -----------------------------
# 2️⃣ Bar Plot (Profit)
# -----------------------------
axs[1].bar(months, profit)

axs[1].set_title("Monthly Profit")
axs[1].set_xlabel("Months")
axs[1].set_ylabel("Amount")

axs[1].grid(axis='y', linestyle='--', alpha=0.6)

# -----------------------------
# 3️⃣ Pie Chart (Product Distribution)
# -----------------------------
axs[2].pie(
    distribution,
    labels=categories,
    autopct='%1.1f%%',
    startangle=90
)

axs[2].set_title("Product Distribution")

# -----------------------------
# 4️⃣ Histogram (Profit Variation)
# -----------------------------
axs[3].hist(profit, bins=5, edgecolor='black')

axs[3].set_title("Profit Distribution")
axs[3].set_xlabel("Profit")
axs[3].set_ylabel("Frequency")

axs[3].grid(True, linestyle='--', alpha=0.6)

# -----------------------------
# Layout Adjustment
# -----------------------------
plt.tight_layout()

plt.show()
````
