# **Temperature Analysis**  
**Goal:** Simulate 30 days of temperature and analyze.  
**Requirements:**
* Line plot
* Scatter
* Histogram
* Subplot
* Mean Line  




````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
days = np.arange(1,31) # Creat array of a month
temperature = np.random.randint(20, 40, size=30) # Creat array of 30 days temperature
temperature
````

## **Line Plot**  

````python
x = days
y = temperature

dict_name = dict(
    marker = '.'
)

plt.plot(x,y, **dict_name)
plt.grid()
plt.yticks(np.arange(20,40,5)) # controling the ticks (start, end, interval)
plt.xticks(np.arange(0,31,5))
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Trend")
plt.show()
````

## **Scatter Plot**  

````python
x = days
y = temperature

dict_name = dict(
    marker = '.',
    color = 'red',
    s = 100,
    label = 'Temperature'
)

plt.scatter(x,y, **dict_name)
plt.grid()
plt.yticks(np.arange(20,40,5)) # controling the ticks (start, end, interval)
plt.xticks(np.arange(0,31,5))
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Trend")

plt.show()
````

## **Histogram**  

````python
x = days
y = temperature

plt.hist(y, bins=10)
plt.grid()
plt.yticks(np.arange(0,10,1)) # controling the ticks (start, end, interval)
plt.xticks(np.arange(20,40,2))
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Temperature Distribution")

plt.show()
````

## **Line plot + horizontal mean line**  

````python
mean = np.mean(y)
mean
````

````python
x = days
y = temperature

dict_name = dict(
    marker = '.'
)

plt.plot(x,y, **dict_name)
plt.axhline(mean, color='r', linestyle='--')
plt.grid()
plt.yticks(np.arange(20,40,5)) # controling the ticks (start, end, interval)
plt.xticks(np.arange(0,31,5))
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Temperature With Mean Line")
plt.show()
````

# **Mathematical Function Dashboard**  
**Goal**  
Build a function visualization dashboard.  
**Requirements**  
Plot:
* Linear Function
* Quadratic Function
* Exponential
* Logarithmic

All in one figure.

**Data**  
```
x = np.linspace(1, 10, 400)
y1 = 2*x + 1
y2 = x**2
y3 = np.exp(x/5)
y4 = np.log(x)
```
**What You Must Do**  
1. 2×2 subplot layout
2. Use different colors
3. Add grid
4. Add minor ticks
5. Share x-axis where appropriate
6. Set axis limits properly

````python
# Creat Dataset
x = np.linspace(1, 10, 40)
y1 = 2*x + 1
y2 = x**2
y3 = np.exp(x/5)
y4 = np.log(x)

# Print values
# print(x)
# print(y1)
# print(y2)
# print(y3)
# print(y4)
# creat 2x2 subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Linear Function
axs[0, 0].plot(x, y1, color='b')
axs[0, 0].set_title('Linear Function')
axs[0, 0].set_xticks(np.arange(1, 11, 1))
axs[0, 0].set_yticks(np.arange(0, 22, 2))
axs[0, 0].minorticks_on()
axs[0, 0].grid(True, which='both')

# Quadratic Function
axs[0, 1].plot(x, y2, color='g')
axs[0, 1].set_title('Qudratic Function')
axs[0, 1].set_xticks(np.arange(0, 11, 1))
axs[0, 1].set_yticks(np.arange(0, 101, 10))
axs[0, 1].minorticks_on()
axs[0, 1].grid(True, which='both')

# Exponential Function
axs[1, 0].plot(x, y3, color='r')
axs[1, 0].set_title('Exponential Function')
axs[1, 0].set_xticks(np.arange(0, 11, 1))
axs[1, 0].set_yticks(np.arange(0, 11, 1))
axs[1, 0].minorticks_on()
axs[1, 0].grid(True, which='both')

# Logarithmic Function
axs[1, 1].plot(x, y4, color='y')
axs[1, 1].set_title('Logarithmic Function')
axs[1, 1].set_xticks(np.arange(1, 11, 1))
axs[1, 1].set_yticks(np.arange(0, 2.5, 0.2))
axs[1, 1].minorticks_on()
axs[1, 1].grid(True, which='both')

plt.tight_layout()
plt.show()
````

# **Mini Data Dashboard**  
**Goal**  
Create a professional data dashboard.  
Simulate a small company dataset:  
* Monthly sales
* Monthly expenses
* Profit
* Category distribution  

**Data Example**  

months = np.array(["Jan","Feb","Mar","Apr","May","Jun"])  
sales = np.array([200, 220, 250, 240, 260, 300])  
expenses = np.array([150, 160, 180, 170, 200, 210])  
profit = sales - expenses  
categories = ["Product A", "Product B", "Product C"]  
distribution = np.array([50, 30, 20])  

**What You Must Do**  
Create ONE big figure with:  
1. Line plot → Sales & Expenses
2. Bar plot → Profit
3. Pie chart → Product distribution
4. Histogram → Profit variation  

**Add:**  
* Legend
* Grid
* Proper layout
* Titles
* Clean style  

Make it look like a business report.

````python
# Dataset
months = np.array(["Jan","Feb","Mar","Apr","May","Jun"])
sales = np.array([200, 220, 250, 240, 260, 300])
expenses = np.array([150, 160, 180, 170, 200, 210])
profit = sales - expenses
categories = ["Product A", "Product B", "Product C"]
distribution = np.array([50, 30, 20])

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line plot - sales & expenses
axs[0, 0].plot(months, sales, label='Sales')
axs[0, 0].plot(months, expenses, label='Expenses')
axs[0, 0].set_title("Line plot - sales & expenses")
axs[0, 0].set_xlabel("Months")
axs[0, 0].set_ylabel("Amount")
axs[0, 0].grid(True)
axs[0, 0].legend()

# Bar plot → Profit
axs[0, 1].bar(months, profit, label='Profit')
axs[0, 1].set_title("Bar plot → Profit")
axs[0, 1].set_xlabel("Months")
axs[0, 1].set_ylabel("Amount")
# axs[0, 1].grid(True)
axs[0, 1].legend()

# Pie chart → Product distribution
axs[1, 0].pie(distribution, labels=categories, autopct='%1.1f%%', startangle=90)
axs[1, 0].set_title("Pie chart → Product distribution")

# Histogram → Profit variation
axs[1, 1].hist(profit, bins=5, edgecolor='black')
axs[1, 1].set_title("Histogram → Profit variation")
axs[1, 1].set_xlabel("Profit")
axs[1, 1].set_ylabel("Frequency")



plt.tight_layout()
plt.show()
````
