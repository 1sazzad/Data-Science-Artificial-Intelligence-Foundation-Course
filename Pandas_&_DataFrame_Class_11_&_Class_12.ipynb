## **Example**

````python
## Without Pandas

data = [
["A",25,40000],
["B",30,50000],
["C",28,45000]
]

total = 0
for row in data:
    total += row[2]

avg = total / len(data)
print(avg)
````

All these operations use indexing and selection.

````python
# Create a dataset
import pandas as pd

data = {
"Name":["Alice","Bob","Charlie","David"],
"Age":[25,30,28,35],
"Salary":[40000,50000,45000,60000],
"City":["Dhaka","Delhi","Dhaka","London"]
}

df = pd.DataFrame(data)

df
````

##  **Selecting Column**

````python
# Select Single Column
print("="*50)
print('Printing Single column')
print(df['Age'])
# Select Multiple Column
print("="*50)
print('Printing Multiple column')
print(df[['Age','Salary']])
````

## **Selecting Rows**  
There are two main methods.
| Method | Meaning                 |
| ------ | ----------------------- |
| `loc`    | label-based indexing    |
| `iloc`   | position-based indexing |



### **`.loc[]` — Label Based Selection**   
```
df.loc[row_label , column_label]
```

````python
# ==========================================================
# Selecting Rows using .loc
# ==========================================================

# Select single row
print("🔹 Select Single Row")
print("=" * 40)
print(df.loc[0])


# Select row and column
print("\n🔹 Select Row and Column")
print("=" * 40)
print(df.loc[0, "Age"])


# Select multiple rows
print("\n🔹 Select Multiple Rows")
print("=" * 40)
print(df.loc[0:2])
````

### **`.iloc[]` — Position Based Selection**  
```
df.iloc[row_position , column_position]
```

````python
# ==========================================================
# Selecting Rows using .iloc
# ==========================================================

# Select single row
print("🔹 Select Single Row")
print("=" * 40)
print(df.iloc[0])


# Select specific cell
print("\n🔹 Select Specific Cell")
print("=" * 40)
print(df.iloc[0, 1])


# Select multiple rows
print("\n🔹 Select Multiple Rows")
print("=" * 40)
print(df.iloc[0:3])
````

### **Difference Between `.loc` and `.iloc`**  
| Feature           | loc           | iloc             |
| ----------------- | ------------- | ---------------- |
| Index type        | label         | position         |
| Uses column names | yes           | no               |
| Uses numbers      | label numbers | position numbers |


## **Conditional Selection (Very Important)**
**Symbols**  
| Symbol | Meaning |
| ------ | ------- |
| &      | AND     |
| |      | OR      |
| ~      | NOT     |


````python
# ==========================================================
# Conditional Selection (Filtering Data)
# ==========================================================

# Select people with age > 28
print("🔹 People with Age > 28")
print("=" * 40)
print(df[df["Age"] > 28])


# Multiple conditions
print("\n🔹 People with Age > 25 and City = Dhaka")
print("=" * 40)
print(df[(df["Age"] > 25) & (df["City"] == "Dhaka")])


# ==========================================================
# Selecting Specific Columns After Filtering
# ==========================================================

# People with age > 28 but show only Name
print("\n🔹 People with Age > 28 (Only Name)")
print("=" * 40)
print(df[df["Age"] > 28]["Name"])


# Selecting rows and columns together
print("\n🔹 People with Age > 28 (Name and Age)")
print("=" * 40)
print(df.loc[df["Age"] > 28, ["Name", "Age"]])
````

# **Combining Data in Pandas**  
In real data science projects, data rarely comes in one single table.  
Usually you have multiple datasets like:  
**Customers Table**  
| CustomerID | Name  |
| ---------- | ----- |
| 1          | Alice |

````python
# Create dataset
import pandas as pd

df1 = pd.DataFrame({
"Name":["Alice","Bob"],
"Age":[25,30]
})

df2 = pd.DataFrame({
"Name":["Charlie","David"],
"Age":[28,35]
})
````

````python
# df1
df1
````

## **Pandas Dataset Combination Functions**  
| Function | Syntax | Function Details | Example |
|---|---|---|---|
| `concat()` | `pd.concat([df1, df2])` | Combines multiple DataFrames.<br>Stacks datasets together.<br>Default operation is row-wise. | `pd.concat([df1, df2])` |
| `concat(axis=1)` | `pd.concat([df1, df2], axis=1)` | Combines DataFrames column-wise.<br>Instead of stacking rows, columns are merged side-by-side. | `pd.concat([df1, df2], axis=1)` |
| `merge()` | `pd.merge(df1, df2, on="column")` | Combines datasets similar to SQL JOIN.<br>Uses a common column as a key.<br>Most commonly used method in data science. | `pd.merge(customers, orders, on="CustomerID")` |
| `merge(inner)` | `pd.merge(df1, df2, on="col", how="inner")` | Returns only matching rows.<br>Records must exist in both datasets. | `pd.merge(customers, orders, on="CustomerID", how="inner")` |
| `merge(left)` | `pd.merge(df1, df2, on="col", how="left")` | Keeps all rows from the left table.<br>Adds matching rows from the right table.<br>Missing matches become `NaN`. | `pd.merge(customers, orders, on="CustomerID", how="left")` |
| `merge(right)` | `pd.merge(df1, df2, on="col", how="right")` | Keeps all rows from the right table.<br>Adds matching rows from the left table. | `pd.merge(customers, orders, on="CustomerID", how="right")` |
| `merge(outer)` | `pd.merge(df1, df2, on="col", how="outer")` | Returns all rows from both tables.<br>Unmatched values are filled with `NaN`. | `pd.merge(customers, orders, on="CustomerID", how="outer")` |
| `join()` | `df1.join(df2)` | Combines DataFrames using index.<br>Less flexible than `merge()`.<br>Often used when indexes already match. | `df1.join(df2)` |

# **Advanced Pandas Operations**  
**We will study these powerful tools:**  
`pivot_table()` `crosstab()` `multi-index` `stack()` `unstack()`

````python
# Dataset
import pandas as pd

data = {
"City":["Dhaka","Dhaka","Delhi","Delhi","London","London"],
"Product":["Laptop","Phone","Laptop","Tablet","Phone","Tablet"],
"Sales":[2000,1500,1800,1200,1600,900]
}

df = pd.DataFrame(data)

df
````

## **Heading**  

| Function | Syntax | Function Details | Example |
|---|---|---|---|
| `pivot_table()` | `pd.pivot_table(df, values, index)` | Creates a summary table from a dataset.<br>Similar to Excel Pivot Tables.<br>Default aggregation is **mean**. | `pd.pivot_table(df, values="Sales", index="City")` |
| `pivot_table(sum)` | `pd.pivot_table(df, values, index, aggfunc="sum")` | Changes aggregation function.<br>Useful for totals instead of averages. | `pd.pivot_table(df, values="Sales", index="City", aggfunc="sum")` |
| `pivot_table(2D)` | `pd.pivot_table(df, values, index, columns)` | Creates a 2-dimensional pivot table.<br>Rows and columns represent categories. | `pd.pivot_table(df, values="Sales", index="City", columns="Product", aggfunc="sum")` |
| `crosstab()` | `pd.crosstab(col1, col2)` | Creates a contingency table.<br>Counts frequency of combinations between variables. | `pd.crosstab(df["City"], df["Product"])` |
| `multi-index` | `df.set_index([col1,col2])` | Creates hierarchical indexing.<br>Multiple levels of index for structured data. | `df.set_index(["City","Product"])` |

Time series data means **data indexed by time**.

`to_datetime()` `date_range()` `resample()` `rolling()` `shift()` `datetime_features`



````python
import pandas as pd

data = {
"Date":["2024-01-01","2024-01-02","2024-01-03"],
"Sales":[100,120,90]
}

df = pd.DataFrame(data)
````

## **Heading**  

| Function | Syntax | Function Details | Example |
|---|---|---|---|
| `to_datetime()` | `pd.to_datetime(column)` | Converts date strings into datetime format.<br>Enables powerful time-based operations in Pandas. | `df["Date"] = pd.to_datetime(df["Date"])` |
| `set_index()` | `df.set_index("Date")` | Sets date column as index.<br>Required for most time-series analysis operations. | `df.set_index("Date", inplace=True)` |
| `date_range()` | `pd.date_range(start, periods)` | Generates sequences of dates automatically.<br>Useful for creating time series datasets. | `pd.date_range(start="2024-01-01", periods=5)` |
| `datetime features` | `df.index.year` `df.index.month` | Extracts date components such as year, month, day.<br>Used for feature engineering in ML. | `df.index.month` |
| `resample()` | `df.resample("M")` | Changes time frequency of data.<br>Often used for aggregating daily data to weekly or monthly. | `df.resample("M").sum()` |
| `resample codes` | `"D" "W" "M" "Y"` | Defines resampling frequency.<br>D = daily, W = weekly, M = monthly, Y = yearly. | `df.resample("W").sum()` |
| `rolling()` | `df[column].rolling(window=n)` | Performs moving window calculations.<br>Used for smoothing or trend analysis. | `df["Sales"].rolling(window=3).mean()` |
| `shift()` | `df[column].shift(n)` | Shifts data forward or backward.<br>Common in time-series feature engineering. | `df["Sales"].shift(1)` |
